```python
file = open("/content/GenomeText.txt")

genomeSequence = file.read()
length = len(genomeSequence)

t = np.linspace(0, 4 * np.pi, length)
x = np.cos(t)
y = np.sin(t)
z = np.linspace(0, 5, length)

coordinates = np.column_stack((x,y,z))
fig = plt.figure(figsize=(5,5))
subPlot = fig.add_subplot(projection = '3d')

colors = {
    "A" : "blue",
    "T" : "red",
    "G" : "green",
    "C" : "cyan"
}

for i,char in enumerate(genomeSequence):
  subPlot.scatter(coordinates[i, 0], coordinates[i, 1], coordinates[i, 2], color = colors[genomeSequence[i]],marker = "o")

subPlot.set_xlabel("X")
subPlot.set_ylabel("Y")
subPlot.set_zlabel("Z")
subPlot.set_title("Genome Sequence")
plt.show()
```

```python
image = mpImg.imread("/content/Random.jfif")
plt.subplot(2,2,1)
plt.imshow(image)

rotatedImage = np.rot90(image)
plt.subplot(2,2,2)
plt.imshow(rotatedImage)

flippedImage = np.fliplr(image)
plt.subplot(2,2,3)
plt.imshow(flippedImage)

grayScaleImage = np.dot(image[..., :3], [0.299, 0.587, 0.1140])
plt.subplot(2,2,4)
plt.imshow(grayScaleImage, cmap = "gray")
```
